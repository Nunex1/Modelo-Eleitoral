```python

import pandas as pd
from zipfile import ZipFile

arquivo_zip = "votacao_candidato_munzona_2026.zip"
arquivo_csv = "votacao_candidato_munzona_2026_BR.csv"

colunas = [
    "NR_TURNO",
    "SG_UF",
    "CD_MUNICIPIO",
    "NM_MUNICIPIO",
    "DS_CARGO",
    "NR_CANDIDATO",
    "NM_URNA_CANDIDATO",
    "SG_PARTIDO",
    "QT_VOTOS_NOMINAIS_VALIDOS"
]

grupo = [
    "SG_UF",
    "CD_MUNICIPIO",
    "NM_MUNICIPIO",
    "NR_CANDIDATO",
    "NM_URNA_CANDIDATO",
    "SG_PARTIDO"
]

parciais = []

with ZipFile(arquivo_zip) as z:
    with z.open(arquivo_csv) as f:
        for bloco in pd.read_csv(
            f,
            sep=";",
            encoding="latin1",
            usecols=colunas,
            chunksize=100_000
        ):
            filtro = (
                (bloco["NR_TURNO"] == 1) &
                (bloco["DS_CARGO"].str.upper() == "PRESIDENTE")
            )

            dados = bloco.loc[filtro]

            parcial = (
                dados.groupby(grupo, as_index=False, dropna=False)
                ["QT_VOTOS_NOMINAIS_VALIDOS"].sum()
            )

            parciais.append(parcial)

resultado = pd.concat(parciais, ignore_index=True)

resultado = (
    resultado.groupby(grupo, as_index=False, dropna=False)
    ["QT_VOTOS_NOMINAIS_VALIDOS"].sum()
)

resultado = resultado.rename(
    columns={"QT_VOTOS_NOMINAIS_VALIDOS": "VOTOS"}
)

resultado["PERCENTUAL"] = (
    100 * resultado["VOTOS"] /
    resultado.groupby("CD_MUNICIPIO")["VOTOS"].transform("sum")
)

display(resultado.head(10))
```

```python

resultado.to_csv(
    "resultado_presidencial_2026.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Base eleitoral consolidada e salva!")
print("Municípios:", resultado["CD_MUNICIPIO"].nunique())
print("Candidaturas:", resultado["NR_CANDIDATO"].nunique())
```

```python

import numpy as np

# Identificar os votos válidos de cada localidade
chaves = ["SG_UF", "CD_MUNICIPIO"]

total_local = resultado.groupby(chaves)["VOTOS"].transform("sum")

# Calcular percentual apenas quando existem votos
resultado["PERCENTUAL"] = np.where(
    total_local > 0,
    100 * resultado["VOTOS"] / total_local,
    np.nan
)

# Separar os municípios brasileiros
base_brasil = resultado.loc[
    (resultado["SG_UF"] != "ZZ") &
    (total_local > 0)
].copy()

# Verificar a soma dos percentuais
somas = base_brasil.groupby(chaves)["PERCENTUAL"].sum()

print("Municípios na base:", len(somas))
print("Percentuais inconsistentes:",
      (~np.isclose(somas, 100)).sum())

print("Total de votos na base brasileira:",
      int(base_brasil["VOTOS"].sum()))

# Salvar nova base
base_brasil.to_csv(
    "resultado_presidencial_2026_brasil.csv",
    index=False,
    encoding="utf-8-sig"
)
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Consolidar votos por candidatura
resumo = (
    base_brasil.groupby(
        ["NR_CANDIDATO", "NM_URNA_CANDIDATO", "SG_PARTIDO"],
        as_index=False
    )["VOTOS"].sum()
)

# Calcular participação nacional, excluindo o exterior
resumo["PERCENTUAL"] = (
    resumo["VOTOS"] / resumo["VOTOS"].sum() * 100
)

# Organizar por quantidade de votos
resumo = resumo.sort_values(
    "VOTOS", ascending=False
).reset_index(drop=True)

display(resumo)

# Criar gráfico
fig, ax = plt.subplots(figsize=(11, 6))

dados_grafico = resumo.iloc[::-1]

ax.barh(
    dados_grafico["NM_URNA_CANDIDATO"],
    dados_grafico["PERCENTUAL"]
)

ax.set_title("Eleições Presidenciais 2026 | Primeiro Turno")
ax.set_xlabel("Participação nos votos válidos (%)")
ax.set_ylabel("Candidatura")

plt.tight_layout()
plt.show()
```

```python

# Consolidar os votos por UF e candidatura

votos_uf = (
    base_brasil.groupby(
        ["SG_UF", "NR_CANDIDATO", "NM_URNA_CANDIDATO"],
        as_index=False
    )["VOTOS"].sum()
)

# Calcular o total de votos válidos de cada UF
total_uf = votos_uf.groupby("SG_UF")["VOTOS"].transform("sum")

# Participação percentual dentro da própria UF
votos_uf["PERCENTUAL_UF"] = (
    votos_uf["VOTOS"] / total_uf * 100
)

print("Unidades federativas:", votos_uf["SG_UF"].nunique())

# Conferir se os percentuais somam 100% em cada UF
validacao = votos_uf.groupby("SG_UF")["PERCENTUAL_UF"].sum()

print("Todas as UFs consistentes:",
      (validacao - 100).abs().lt(0.0001).all())

display(votos_uf.head(15))
```

```python

import pandas as pd
from zipfile import ZipFile

arquivo_zip = "perfil_eleitorado_2026.zip"

ufs = {
    "AC", "AL", "AP", "AM", "BA", "CE", "DF",
    "ES", "GO", "MA", "MT", "MS", "MG", "PA",
    "PB", "PR", "PE", "PI", "RJ", "RN", "RS",
    "RO", "RR", "SC", "SP", "SE", "TO"
}

colunas = [
    "SG_UF",
    "CD_MUNICIPIO",
    "NM_MUNICIPIO",
    "DS_GENERO",
    "DS_IDENTIDADE_GENERO",
    "QT_ELEITORES"
]

grupo = colunas[:-1]
partes = []

with ZipFile(arquivo_zip) as z:
    arquivos = sorted([
        nome for nome in z.namelist()
        if nome.upper().endswith(
            tuple(f"_{uf}.CSV" for uf in ufs)
        )
    ])

    print("Arquivos estaduais encontrados:", len(arquivos))

    if len(arquivos) != 27:
        raise ValueError(
            "Esperados 27 arquivos estaduais. "
            "Verifique a lista antes de continuar."
        )

    for arquivo in arquivos:
        with z.open(arquivo) as f:
            for bloco in pd.read_csv(
                f,
                sep=";",
                encoding="latin1",
                usecols=colunas,
                chunksize=100_000
            ):
                parcial = (
                    bloco.groupby(
                        grupo,
                        as_index=False,
                        dropna=False
                    )["QT_ELEITORES"].sum()
                )
                partes.append(parcial)

        print("Processado:", arquivo)

perfil_completo = pd.concat(
    partes, ignore_index=True
)

perfil_completo = (
    perfil_completo.groupby(
        grupo,
        as_index=False,
        dropna=False
    )["QT_ELEITORES"].sum()
)

print("\nUnidades federativas:",
      perfil_completo["SG_UF"].nunique())

print("Municípios:",
      perfil_completo[
          ["SG_UF", "CD_MUNICIPIO"]
      ].drop_duplicates().shape[0])

print("Eleitores cadastrados:",
      int(perfil_completo["QT_ELEITORES"].sum()))

display(perfil_completo.head(10))
```

```python

import pandas as pd
import unicodedata

def normalizar(texto):
    texto = str(texto).upper().strip()
    return "".join(
        c for c in unicodedata.normalize("NFKD", texto)
        if not unicodedata.combining(c)
    )

# Carregar tabela geográfica
geo = pd.read_excel("base cidades completa.xlsx")
geo.columns = geo.columns.str.strip().str.lower()

geo["CHAVE"] = geo["cidade"].map(normalizar)
geo["UF"] = geo["uf"].str.upper().str.strip()

geo["DDD"] = pd.to_numeric(
    geo["ddd"], errors="coerce"
).astype("Int64")

# Municípios da base eleitoral
municipios = base_brasil[
    ["SG_UF", "CD_MUNICIPIO", "NM_MUNICIPIO"]
].drop_duplicates().copy()

municipios["CHAVE"] = municipios["NM_MUNICIPIO"].map(normalizar)

# Verificar duplicações no cadastro geográfico
if geo.duplicated(["UF", "CHAVE"]).any():
    raise ValueError("Existem chaves geográficas duplicadas.")

# Integrar informações
municipios_ddd = municipios.merge(
    geo[["UF", "CHAVE", "DDD", "codigo_ibge"]],
    left_on=["SG_UF", "CHAVE"],
    right_on=["UF", "CHAVE"],
    how="left",
    validate="many_to_one"
)

# Auditar cobertura
sem_ddd = municipios_ddd[municipios_ddd["DDD"].isna()]

print("Municípios eleitorais:", len(municipios))
print("Municípios sem DDD:", len(sem_ddd))
print("DDDs encontrados:", municipios_ddd["DDD"].nunique())

display(sem_ddd[["SG_UF", "NM_MUNICIPIO"]].head(30))
```

```python

# Agrupar categorias de identidade de gênero

perfil_completo["IDENTIDADE_GENERO_AGRUPADA"] = (
    perfil_completo["DS_IDENTIDADE_GENERO"].replace({
        "NÃO INFORMADO": "NÃO INFORMADO / PREFERE NÃO INFORMAR",
        "Prefere não informar": "NÃO INFORMADO / PREFERE NÃO INFORMAR"
    })
)

# Consolidar quantidades de eleitores
perfil_agrupado = (
    perfil_completo.groupby(
        [
            "SG_UF",
            "CD_MUNICIPIO",
            "NM_MUNICIPIO",
            "DS_GENERO",
            "IDENTIDADE_GENERO_AGRUPADA"
        ],
        as_index=False,
        dropna=False
    )["QT_ELEITORES"].sum()
)

# Validar que nenhum eleitor foi perdido
assert (
    perfil_agrupado["QT_ELEITORES"].sum()
    == perfil_completo["QT_ELEITORES"].sum()
)

# Visualizar distribuição nacional
resumo_identidade = (
    perfil_agrupado.groupby(
        "IDENTIDADE_GENERO_AGRUPADA",
        as_index=False
    )["QT_ELEITORES"].sum()
)

resumo_identidade["PERCENTUAL"] = (
    100 * resumo_identidade["QT_ELEITORES"]
    / resumo_identidade["QT_ELEITORES"].sum()
)

display(resumo_identidade)
```

```python

import pandas as pd

# Consolidar eleitores por município e gênero cadastral
base_genero = perfil_completo.copy()

base_genero["DS_GENERO"] = (
    base_genero["DS_GENERO"]
    .str.upper()
    .str.strip()
)

base_genero["GRUPO"] = (
    base_genero["DS_GENERO"]
    .map({
        "MASCULINO": "HOMENS",
        "FEMININO": "MULHERES"
    })
    .fillna("OUTROS / NÃO INFORMADO")
)

resumo_genero = base_genero.pivot_table(
    index=["SG_UF", "CD_MUNICIPIO", "NM_MUNICIPIO"],
    columns="GRUPO",
    values="QT_ELEITORES",
    aggfunc="sum",
    fill_value=0
).reset_index()

for coluna in ["HOMENS", "MULHERES", "OUTROS / NÃO INFORMADO"]:
    if coluna not in resumo_genero.columns:
        resumo_genero[coluna] = 0

resumo_genero["TOTAL_ELEITORES"] = (
    resumo_genero["HOMENS"]
    + resumo_genero["MULHERES"]
    + resumo_genero["OUTROS / NÃO INFORMADO"]
)

resumo_genero["PCT_HOMENS"] = (
    100 * resumo_genero["HOMENS"]
    / resumo_genero["TOTAL_ELEITORES"]
)

resumo_genero["PCT_MULHERES"] = (
    100 * resumo_genero["MULHERES"]
    / resumo_genero["TOTAL_ELEITORES"]
)

# Exibir valores no padrão brasileiro
formatacao = {
    coluna: lambda x: f"{x:,.0f}".replace(",", ".")
    for coluna in [
        "HOMENS", "MULHERES",
        "OUTROS / NÃO INFORMADO", "TOTAL_ELEITORES"
    ]
}

formatacao.update({
    "PCT_HOMENS": lambda x: f"{x:.1f}%".replace(".", ","),
    "PCT_MULHERES": lambda x: f"{x:.1f}%".replace(".", ",")
})

display(
    resumo_genero.head(20).style.format(formatacao)
)
```
